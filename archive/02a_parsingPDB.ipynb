```python
#Parse a PDB file and print the chain IDs
from Bio.PDB import PDBParser


# get the Amino Acid Composition of a chain
def get_aa_composition(chain):
    aa_count = {}
    for residue in chain:
        res_name = residue.get_resname()
        if res_name not in aa_count:
            aa_count[res_name] = 0
        aa_count[res_name] += 1
    return aa_count

# show the amino acid composition of each chain in the structure as chart
for model in structure:
    for chain in model:
        aa_composition = get_aa_composition(chain)
        print(f"Chain {chain.id} amino acid composition: {aa_composition}")
```

```python
from Bio.PDB import PDBParser, PPBuilder

parser = PDBParser(QUIET=True)
structure = parser.get_structure("6xc3", "../data/pdb_cnf/6xc3.pdb")

# get cdr h3 region, H3 lies between residues 95 and 102 in the heavy chain
cdr_h3_residues = []

for model in structure:
    for chain in model:
        if chain.id == "H":  # heavy chain
            for res in chain:
                res_id = res.get_id()
                if res_id[0] == " " and 95 <= res_id[1] <= 102:  # Chothia H3
                    cdr_h3_residues.append(res)

# print residue names or sequence
from Bio.Seq import Seq
from Bio.SeqUtils import seq1

aa_seq = ''.join(seq1(res["CA"].get_parent().resname) for res in cdr_h3_residues if "CA" in res)
print("CDR-H3 sequence:", aa_seq)

```
