### Imports and config 

```python
import os
import requests
import pandas as pd
```

### Create folder for Chotia numbered homo sapiens PDB files

```python
PDB_URL = "https://opig.stats.ox.ac.uk/webapps/sabdab-sabpred/sabdab/pdb/{}/?scheme=chothia"
PDB_DIR3 = os.path.abspath(os.path.join(os.getcwd(), "..", "data", "pdbs_h.sapiens_cnf"))
CSV_FILENAME3 = "pdb_ids_homo_sapiens.csv"
os.makedirs(PDB_DIR3, exist_ok=True)

csv_path = os.path.join(os.path.dirname(__file__) if '__file__' in globals() else os.getcwd(), CSV_FILENAME3)
df = pd.read_csv(csv_path, header=None)

unique_pdb_ids = df[0].dropna().astype(str).str.strip()
unique_pdb_ids = unique_pdb_ids[unique_pdb_ids.str.match(r"^[A-Za-z0-9]{4}$")].tolist()

print(f"{len(unique_pdb_ids)} valid PDB-IDs loaded.")
print(f"File saved in: {PDB_DIR3}")
```

### Download function

```python
def download_pdb(pdb_id):
    filename = os.path.join(PDB_DIR3, f"{pdb_id}.pdb")

    # Check if file already exists in folder - skip download
    if os.path.exists(filename) > 100:
        print(f"{pdb_id} already exists - download skipped.")
        return filename
    else: 
        print(f"{pdb_id} to be downloaded.")

    # Download files that are still missing
    url = PDB_URL.format(pdb_id)
    try: 
        response = requests.get(url, timeout=10)
        if response.status_code == 200 and len(response.content) > 100:
            with open(filename, "wb") as f: 
                f.write(response.content)
            print(f"{pdb_id} downloaded successfully.")
            return filename
        else:
            print(f"{pdb_id} could not be downloaded (invalid content or status code).")
            return None
    except requests.exceptions.RequestException as e:
        print(f"{pdb_id} download failed due to connection error: {e}")
        return None

```

### Start download

```python
print(f"Start Downloading {len(unique_pdb_ids)} PDB-Files...\n")

for i, pdb_id in enumerate(unique_pdb_ids, 1):
    print(f"[{i}/{len(unique_pdb_ids)}] {pdb_id}:", end=" ")
    download_pdb(pdb_id)

print("\nDone.")
```
