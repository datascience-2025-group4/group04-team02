# Download Chotia renumbered PDB files from SAbDaB
### SAbDaB provides a webervice to download Chotia-renumbered PDB files at the following URL


```python
PDB_URL = "https://opig.stats.ox.ac.uk/webapps/sabdab-sabpred/sabdab/pdb/{}/?scheme=chothia"
```

## Create function to download PDB file
### Create the URL to download the PDB file for single pdb file - example pdb_id '9ds2'


```python
pdb_id = '9ds2'

url = PDB_URL.format(pdb_id)

url
```

### To download files from the internet, python provides the `requests`library.

### The function `request.get(URL)`returnst a response object with attributes
### - status_code
### - content

### to ensure everything goes as planned when downloading files from the internet, it is important to check the `status_code`. A value of 200 means success. Any other value means something went wrong. 

```python
import requests 

response = requests.get(url)

response.status_code
```

Now look at the first 100 characters of the `content` attribute

```python
response.content[0:100]
```
