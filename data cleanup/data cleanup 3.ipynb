### This function 'download_pdb(pdb_id) will summarise and execute all the steps above.

### Filename will be returned if the download was successful, otherwise 'None'

### Step overwiev
### - create URL from PDB_URL template
### - create filename
### - if file excists, return filename (this allows us to return to add failed files)
### - download data 
### - check status_code. If status_code == 200 and content lenght > 100, 
### --> write content 
### --> return filename

### - ortherwise, 
### - print an error message stating thet pdb_id could not be downloaded
### - return None

```python
import os

def download_pdb(pdb_id):
    url = PDB_URL.format(pdb_id)
    filename = f'{PDB_DIR}/{pdb_id}'
    if os.path.exists(filename):
        return(filename)
    else: 
        response = requests.get(url)
        if response.status_code == 200 and len(response.content) > 100:
            with open(filename,"wb") as f:
                f.write(response.content)
                return(filename)
        else:
            print(f"{PDB_DIR} could not be downloaded")
            return None
```
