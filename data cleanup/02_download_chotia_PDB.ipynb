## Download Chothia renumbered PDB files from SAbDaB



### Imports and config 

```python
import os
import requests
import pandas as pd
```

### Load Paths and CSV's 

```python
DATASETS = [
    ("../data/pdbs_corona_cnf", "pdb_ids_sars_cov2.csv"),
    ("../data/pdbs_h.sapiens_cnf", "pdb_ids_homo_sapiens.csv"),
    ("../data/pdbs_influenza_cnf", "pdb_ids_influenza.csv"),
]
```

### Download function

```python
PDB_URL = "https://opig.stats.ox.ac.uk/webapps/sabdab-sabpred/sabdab/pdb/{}/?scheme=chothia"

def download_pdb(pdb_id, pdb_dir):
    filename = os.path.join(pdb_dir, f"{pdb_id}.pdb")
#def download_pdb(pdb_id):
    #filename = os.path.join(PDB_DIR, f"{pdb_id}.pdb")

    # Wenn Datei schon existiert → überspringen
    if os.path.exists(filename):
        print(f"{pdb_id} already exists – download skipped.")
        return filename

    # Herunterladen
    url = PDB_URL.format(pdb_id)
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200 and len(response.content) > 100:
            with open(filename, "wb") as f:
                f.write(response.content)
            print(f"{pdb_id} downloaded successfully.")
            return filename
        else:
            print(f"{pdb_id} could not be downloaded (invalid content or status code).")
            return None
    except requests.exceptions.RequestException as e:
        print(f"{pdb_id} download failed due to connection error: {e}")
        return None
```

```python
def download_batch(csv_filename, pdb_dir):
    os.makedirs(pdb_dir, exist_ok=True)
    df = pd.read_csv(csv_filename)
    unique_pdb_ids = df["pdb"].unique()

    for i, pdb_id in enumerate(unique_pdb_ids, 1):
        print(f"[{i}/{len(unique_pdb_ids)}] {pdb_id}:", end=" ")
        download_pdb(pdb_id)

    print("\nDone.")

    for csv_filename, pdb_dir in DATASETS:
        download_batch(csv_filename, pdb_dir)
```

### Download Data

```python
for csv_filename, pdb_dir in DATASETS:
    download_batch(csv_filename, pdb_dir)
```
