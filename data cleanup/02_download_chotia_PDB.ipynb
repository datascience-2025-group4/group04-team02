## Download Chothia renumbered PDB files from SAbDaB



### Imports and config 

```python
import os
import requests
import pandas as pd
```

```python
df = pd.read_csv("../generated/data cleanup/ab_ag_filtered_pdb.tsv", sep="\t")
```

mapping of code 

```python
DATASETS = [
    ("SARS-CoV-2", "../data/pdbs_corona_cnf"),   
    ("Influenza A", "../data/pdbs_influenza_cnf"),
    ("Homo Sapiens", "../data/pdbs_h.sapiens_cnf")
]
```

### Download function

```python
PDB_URL = "https://opig.stats.ox.ac.uk/webapps/sabdab-sabpred/sabdab/pdb/{}/?scheme=chothia"

def download_pdb(pdb_id, pdb_dir):
    filename = os.path.join(pdb_dir, f"{pdb_id}.pdb")


    # If file already exists -> skip download 
    if os.path.exists(filename):
        print(f"{pdb_id} already exists – download skipped.")
        return filename

    # HerDownload
    url = PDB_URL.format(pdb_id)
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200 and len(response.content) > 100:
            with open(filename, "wb") as f:
                f.write(response.content)
            print(f"{pdb_id} downloaded successfully.")
            return filename
        else:
            print(f"{pdb_id} could not be downloaded (invalid content or status code).")
            return None
    except requests.exceptions.RequestException as e:
        print(f"{pdb_id} download failed due to connection error: {e}")
        return None
```

```python
def download_batch(pdb_ids, pdb_dir):
    os.makedirs(pdb_dir, exist_ok=True)

    for i, pdb_id in enumerate(pdb_ids, 1):
        print(f"[{i}/{len(pdb_ids)}] {pdb_id}:", end=" ")
        download_pdb(pdb_id, pdb_dir)

    print("\nDone.")
```

### Download Data

```python
pdb_ids = df["pdb"].tolist()
download_batch(pdb_ids, "../data/pdbs_all_cnf")
```

Files also Downloaded seperately for each species, as required by some notebooks

```python
for species, pdb_dir in DATASETS:
    pdb_ids = df.query("species == @species")["pdb"].tolist()
    download_batch(pdb_ids, pdb_dir)
```
