### Bedeutung von (het_flag, residue_number, insertion_code)

### het_flag (Hetero-Flag):
- Gibt an, ob es sich um eine normale Aminosäure oder um etwas anderes handelt.
- Wenn es eine normale Aminosäure ist, ist het_flag ein Leerzeichen ' ' (Achtung: nicht leer, sondern ein tatsächliches Leerzeichen!).
- Wenn es ein Hetero-Atom ist, steht dort z. B.: 'W' für Wasser, 'H_GLC' für Glukose ...

### residue_number (Residuenummer):
- Die Position des Residues in der Kette (z. B. 95, 97...).
- Wenn du Chothia-nummerierte PDB-Dateien verwendest, ist das die Chothia-Position

### insertion_code:
- Wird benutzt, wenn es Einschübe bei der Nummerierung gibt (z. B. "97A", "97B").
- Wenn kein Einschub vorliegt, ist auch das ein Leerzeichen ' '.

Consider pdb_id '9ds2'

- create the path to the file
- parse the file
- retrieve all residues of the 'H' chain (this can be done in a single line in python)
- filter on amino acid residues (i.e. the `het_flag` of the parent residue is ' ', space not empty).
  Here one can use a list comprehension in python:
  `[res for res in residues if res.id[0] == ' ']`


### Zile 
#### - Aus allen PDB-Dateien im Ordner die schwere Kette 'H' auslesen
#### - Alle Residuen dieser Kette holen
#### - Nur echte Aminosäuren behalten (filter nach het_flag == ' ' = Leerzeichen)
#### - Ergebnis für jede PDB speichern (z.B. als Liste von Residuen oder als Aminosäure-Sequenz)

#### = Ziel ist es, mehrere PDBs gleichzeitig zu parsen und echte Aminosäure-Residuen zu extrahieren


```python
import os
from Bio.PDB import PDBParser

# 1. Ordner, in dem die PDB-Dateien liegen
PDB_DIR = "../data/pdbs"

# 2. PDBParser initialisieren
parser = PDBParser(PERMISSIVE=1)

# 3. Ergebnis-Dictionary, um für jede PDB die Aminosäure-Residuen zu speichern
results = {}

# 4. Alle Dateien im Ordner durchgehen
for file in os.listdir(PDB_DIR):
    if not file.endswith(".pdb"):  # nur PDB-Dateien bearbeiten
        continue

    pdb_id = file.split("_")[0]  # z.B. '9ds2_chothia.pdb' -> '9ds2'
    filepath = os.path.join(PDB_DIR, file)

    try:
        # 5. PDB-Struktur laden
        structure = parser.get_structure(pdb_id, filepath)

        # 6. Erstes Modell auswählen (bei NMR-Strukturen kann es mehrere Modelle geben)
        model = structure[0]

        # 7. Prüfen, ob Kette 'H' vorhanden ist
        if 'H' not in model:
            print(f"{pdb_id}: Kette 'H' nicht gefunden, übersprungen")
            continue

        # 8. Kette 'H' holen
        chainH = model['H']

        # 9. Alle Residuen der Kette holen (als Liste)
        residues = list(chainH.get_residues())

        # 10. Filter: Nur echte Aminosäuren behalten (het_flag ist Leerzeichen ' ')
        amino_acid_residues = [res for res in residues if res.id[0] == ' ']

        # 11. Ergebnis abspeichern
        results[pdb_id] = amino_acid_residues

    except Exception as e:
        print(f"Fehler bei {file}: {e}")

# 12. Ausgabe zur Kontrolle (PDB und Anzahl gefilterter Aminosäuren)
for pdb, res_list in results.items():
    print(f"{pdb}: {len(res_list)} echte Aminosäure-Residuen in Kette 'H'")
```

### code wenn länge der sequenz mit limit begrenzt werden soll

### ziel:
- Code liest mehrere PDB-Dateien + sucht in jeder Datei die schwere Kette ('H') heraus
- Aminosäuresequenz wir erstellt — aber nur bis zu einer bestimmten Position (Residuum-Nummer limit).

- Dabei wird aus den 3-Buchstaben-Aminosäure-Codes (z.B. 'ALA' für Alanin) der gebräuchliche 1-Buchstaben-Code (z.B. 'A') gemacht. Falls eine Aminosäure unbekannt ist, wird ein 'X' eingefügt

```python
import os
from Bio.PDB import PDBParser

# Dictionary für 3-Buchstaben- zu 1-Buchstaben-Code
three2one = {
    'ALA': 'A', 'ARG': 'R', 'ASN': 'N', 'ASP': 'D',
    'CYS': 'C', 'GLN': 'Q', 'GLU': 'E', 'GLY': 'G',
    'HIS': 'H', 'ILE': 'I', 'LEU': 'L', 'LYS': 'K',
    'MET': 'M', 'PHE': 'F', 'PRO': 'P', 'SER': 'S',
    'THR': 'T', 'TRP': 'W', 'TYR': 'Y', 'VAL': 'V'
}

PDB_DIR = "../data/pdbs"  # Pfad zum Ordner mit PDB-Dateien
parser = PDBParser(PERMISSIVE=1)  # Parser initialisieren

def extract_aa_sequence(pdb_id, chain, limit):
    """
    Extrahiert die Aminosäure-Sequenz der angegebenen Kette bis zum Residuum 'limit' aus einer PDB-Datei.

    Args:
        pdb_id (str): ID der PDB-Datei (z.B. '9ds2')
        chain (str): Kettenbezeichnung (z.B. 'H')
        limit (int): Maximale Residuum-Nummer (Residue Number)

    Returns:
        str: Aminosäuresequenz als Ein-Buchstaben-Code-String
    """
    # 1. Pfad zur PDB-Datei erstellen
    filepath = os.path.join(PDB_DIR, f"{pdb_id}_chothia.pdb")

    # 2. PDB-Datei parsen
    structure = parser.get_structure(pdb_id, filepath)

    # 3. Erstes Modell holen (typisch ist nur eins vorhanden)
    model = structure[0]

    # 4. Kette aus Modell holen
    if chain not in model:
        return ""  # Falls Kette nicht existiert, leere Sequenz zurückgeben
    chain_obj = model[chain]

    # 5. Alle echten Aminosäure-Residuen filtern:
    #    - het_flag == ' ' (echte Aminosäure, kein Wasser oder Ligand)
    #    - Residue-Nummer <= limit
    residues = [
        res for res in chain_obj.get_residues()
        if res.id[0] == ' ' and res.id[1] <= limit
    ]

    # 6. 3-Buchstaben-Namen in 1-Buchstaben-Codes umwandeln, 'X' bei unbekannten
    aa_seq = [three2one.get(res.get_resname(), 'X') for res in residues]

    # 7. Liste der Buchstaben zu einem String verbinden
    sequence = "".join(aa_seq)

    # 8. Sequenz zurückgeben
    return sequence


# Beispiel-Anwendung für mehrere PDBs
with open("pdb_ids.txt") as f:
    pdb_list = [line.strip() for line in f if line.strip()]  # Liste aller PDB-IDs aus Datei

for pdb in pdb_list:
    seq = extract_aa_sequence(pdb, 'H', 120)  # Kette 'H', bis Residuum 120
    print(f"PDB {pdb}, Kette H, Sequenz (bis 120): {seq}")
```
