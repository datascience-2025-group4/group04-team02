## relative pathways 

```python
from pathlib import Path

# Absoluter Pfad zum Projektordner (dein Projektverzeichnis)
project_root = Path(r"C:\Users\clara\group04-team02")

# Ordner, in dem die PDB-Dateien liegen
pdb_folder = project_root / "data cleanup" / "pdb_cf"

# Alle .pdb-Dateien finden
pdb_files = list(pdb_folder.glob("*.pdb"))

# Relativ zum Projektordner machen und anzeigen
for pdb in pdb_files:
    print(pdb.relative_to(project_root))

```
