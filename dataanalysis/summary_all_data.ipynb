```python
import pandas as pd

# 1. Pfade als "roh"-Strings (damit auch Leerzeichen richtig interpretiert werden)
df_human = pd.read_csv("../data cleanup/human_cdr_seq.tsv", sep="\t")
df_influenza = pd.read_csv("../data cleanup/influenza_cdr_seq.tsv", sep="\t")
df_sars = pd.read_csv("../data cleanup/sars_cov2_cdr_seq.tsv", sep="\t")

# 2. Quelle hinzufügen
df_human["quelle"] = "human"
df_influenza["quelle"] = "influenza"
df_sars["quelle"] = "sars_cov2"

# 3. Zusammenführen
all_data = pd.concat([df_human, df_influenza, df_sars], ignore_index=True)

# 4. Kontrolle
print("✅ Zusammengeführt:", all_data.shape[0], "Zeilen,", all_data.shape[1], "Spalten")
print("📋 Spaltennamen:", list(all_data.columns)[:10], "...")

# 5. Speichern
all_data.to_csv("all_data.tsv", sep="\t", index=False)
print("💾 Gespeichert als: all_data.tsv")
```
