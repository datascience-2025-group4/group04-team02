```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# coronavirus CDR

```python
data = pd.read_csv('corona.tsv', sep='\t')
```

```python
data
```

```python
for i in range(1,4):
    data[f'len_H{i}'] = data[f'CDR_H{i}'].apply(lambda x: len(x) if isinstance(x, str) else 0)
```

```python
data
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(data['len_H1'], data['len_H2'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H2')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data['len_H2'], data['len_H3'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H2')
plt.ylabel('Length of CDR_H3')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data['len_H1'], data['len_H3'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H3')
plt.show()
```

```python
data[['len_H1', 'len_H2', 'len_H3']]
```

```python
corr = data[['len_H1', 'len_H2', 'len_H3']].corr(method='pearson')
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your file
data = pd.read_csv(r'C:\Users\clara\group04-team02\data cleanup\human_cdr_seq.tsv', sep='\t')

# Compute lengths
data['len_H1'] = data['CDR_H1'].str.len()
data['len_H2'] = data['CDR_H2'].str.len()
data['len_H3'] = data['CDR_H3'].str.len()

# Plot normalized histograms
plt.figure(figsize=(10, 6))
plt.hist(data['len_H1'].dropna(), bins=20, density=True, alpha=0.5, label='CDR_H1')
plt.hist(data['len_H2'].dropna(), bins=20, density=True, alpha=0.5, label='CDR_H2')
plt.hist(data['len_H3'].dropna(), bins=20, density=True, alpha=0.5, label='CDR_H3')

plt.title('Normalized length distribution of CDRs')
plt.xlabel('Length')
plt.ylabel('Relative Frequency (Density)')
plt.legend()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.kdeplot(data['len_H1'].dropna(), label='CDR_H1', fill=True)
sns.kdeplot(data['len_H2'].dropna(), label='CDR_H2', fill=True)
sns.kdeplot(data['len_H3'].dropna(), label='CDR_H3', fill=True)

plt.title('Smoothed length distribution (KDE) of CDRs')
plt.xlabel('Length')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(data['len_H1'], data['len_H2'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H2')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data['len_H2'], data['len_H3'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H2')
plt.ylabel('Length of CDR_H3')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data['len_H1'], data['len_H3'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H3')
plt.show()
```
