## CDR amino acid sequence length 

Analysis of CDR amino acid sequence lengths in antibody regions H1, H2, and H3 across coronavirus, human, and influenza.

This Jupyter notebook aims to characterize the lengths of the CDR regions, compare differences between CDR H1, H2, and H3, and identify any variations among antibodies from coronavirus, human, and influenza sources.
Therefore, the lengths of the CDR regions for each organism were first extracted. Additionally, correlations between the lengths of different regions were analyzed to assess whether, for example, a longer CDR H1 sequence is associated with a longer CDR H2 region.

## Coronavirus

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

For each of the three CDR regions (CDR_H1, CDR_H2, and CDR_H3), the code computed the length of the amino acid sequence. Specifically, for every CDR column, a function was applied that determined the number of residues in the sequence if the value was a string; otherwise, a length of 0 was assigned.
The resulting sequence lengths were stored in three new columns named len_H1, len_H2, and len_H3. Finally, these columns were printed to display the lengths of each CDR region for all records in the dataset.

This code calculates the Spearman correlation coefficients between the lengths of the three CDR regions within each dataset separately (SARS-CoV-2, Human, and Influenza). For each organism, it computes the correlation matrix showing how strongly the lengths of CDR H1, H2, and H3 are related. The heatmaps then display these correlations, making it easy to compare patterns between the datasets and see whether certain CDR lengths tend to vary together.
The Spearman method was used instead of the Pearson method, since it is less sensitive to outliers and usually used when the values are not normally distributed.

```python
datasets = {
    "SARS-CoV-2": data_corona,
    "Human": data_human,
    "Influenza": data_influenza
}

plt.figure(figsize=(15, 5))

for i, (name, df) in enumerate(datasets.items(), 1):
   
    len_cols = [f'len_H{i}' for i in range(1, 4)]
    corr = df[len_cols].corr(method="spearman")

    
    plt.subplot(1, 3, i)
    sns.heatmap(
        corr, 
        annot=True, 
        vmin=-1, vmax=1,
        cmap="coolwarm",
        square=True,
        cbar=False
    )
    plt.title(f"correlation of CDR-lengths\n{name}")

plt.tight_layout()
plt.show() 
```

These heatmap do not show an instant correlation among CDR lengths, however a moderate negative correlation between the lengths of the CDR H1 and CDR H2 regions (r ≈ -0.3 to -0.5) is noticeable. This indicates a tendency among the three organsims, but not a strict inverse correlation.
Additionally, CDR H2 and H3 are moderately correlated in SARS-CoV-2, nevertheless this trait is not present in Human and Influenza CDRs.

In order to get a brief overview, the data is reshaped into a long format so each row represents a single CDR measurement with its source label. Descriptive statistics, including minimum, maximum, mean, standard deviation, and range, are computed for each CDR type in each organism group, allowing an overview of how CDR lengths differ across datasets.

```python
import pandas as pd

# 1) Read TSV files
sars_df = pd.read_csv(r'generated\cdrs\seq\corona_cdr_seq.tsv', sep="\t")
human_df = pd.read_csv(r'generated\cdrs\seq\human_cdr_seq.tsv', sep="\t")
influenza_df = pd.read_csv(r'generated\cdrs\seq\influenza_cdr_seq.tsv', sep="\t")

# 2) Combine all into one DataFrame
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H1_len": df["CDR_H1"].str.len(),
        "CDR_H2_len": df["CDR_H2"].str.len(),
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Transform into long format
lengths_long = df_all.melt(
    id_vars="Source",
    value_vars=["CDR_H1_len", "CDR_H2_len", "CDR_H3_len"],
    var_name="CDR",
    value_name="Length"
)

# Clean up the CDR column to just H1, H2, H3
lengths_long["CDR"] = lengths_long["CDR"].str.replace("_len", "")

# 4) Group by CDR and organism to compute descriptive statistics
stats = lengths_long.groupby(["Source", "CDR"])["Length"].agg(
    Min="min",
    Max="max",
    Mean="mean",
    Std="std",
    Range=lambda x: x.max() - x.min()
)

# 5) Print the results
print("Length statistics per organism and CDR:")
print(stats)
```

This code performs a hierarchical cluster analysis based on the mean lengths of the CDRs. The data is organized into a table containing the mean CDR lengths per organism and region. The table is then reshaped so each organism is represented by a vector of mean lengths across the three CDRs. Using Ward’s linkage method, a hierarchical clustering is computed to group the organisms based on similarity in their CDR length profiles. Finally, the resulting dendrogram visually represents these relationships by displaying clusters and the distances between them.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

# Your data as a DataFrame (recreate from your table)
data = {
    ('Flu', 'CDR_H1'): [7.806122],
    ('Flu', 'CDR_H2'): [5.612245],
    ('Flu', 'CDR_H3'): [17.112245],
    ('Human', 'CDR_H1'): [7.159091],
    ('Human', 'CDR_H2'): [5.906566],
    ('Human', 'CDR_H3'): [11.446970],
    ('SARS', 'CDR_H1'): [7.145610],
    ('SARS', 'CDR_H2'): [5.674518],
    ('SARS', 'CDR_H3'): [13.194861],
}

# Convert to multiindex dataframe
df = pd.DataFrame(data).T.reset_index()
df.columns = ['Organism', 'CDR', 'MeanLength']

# Pivot to have CDR types as columns, organisms as rows
df_pivot = df.pivot(index='Organism', columns='CDR', values='MeanLength')


# Hierarchical clustering on the means
linkage_matrix = linkage(df_pivot, method='ward')

# Plot dendrogram
plt.figure(figsize=(8, 5))
dendrogram(linkage_matrix, labels=df_pivot.index)
plt.title('Hierarchical Clustering of Organisms by Mean CDR Lengths')
plt.ylabel('Distance')
plt.show()
```

The dendrogram shows that Human and SARS cluster closely together, reflecting their similar mean CDR lengths across all three regions, especially in CDR H3 where both have shorter average lengths compared to Flu. Flu forms a separate cluster, which indicates its CDR length profile is distinct, particularly due to its much longer and more variable CDR H3 region. This shows that Flu antibodies differ substantially in their CDR characteristics.

### Statistical hypothesis testing


This code loads CDR sequence data for SARS-CoV-2, human, and influenza antibodies and extracts the lengths of the heavy-chain CDR1, CDR2, and CDR3 regions. 
The code above showed some variances regarding the distribution of the lengths. However, some lengths are minimally represented. To test whether the variances of these lengths differ significantly between CDR types, it is helpful to conduct a statistical hypothesis test.
In this case, the Levene’s test analyzes homogeneity of variance. Levene’s test calculates the distance of each observation from its group mean and tests whether the spread of those distances is the same across groups.
A significant result (p < 0.05) indicates that the variability of lengths is different across the CDRs. Finally, a boxplot visualizes the distributions and the spread of the CDR lengths for easier comparison.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import levene

# 1) Load the data
sars_df = pd.read_csv(r'generated\cdrs\seq\corona_cdr_seq.tsv', sep="\t")
human_df = pd.read_csv(r'generated\cdrs\seq\human_cdr_seq.tsv', sep="\t")
influenza_df = pd.read_csv(r'generated\cdrs\seq\influenza_cdr_seq.tsv', sep="\t")

# 2) Extract CDR lengths
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H1_len": df["CDR_H1"].str.len(),
        "CDR_H2_len": df["CDR_H2"].str.len(),
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Reshape data into long format for easier plotting
lengths_long = df_all.melt(
    id_vars="Source",
    value_vars=["CDR_H1_len", "CDR_H2_len", "CDR_H3_len"],
    var_name="CDR",
    value_name="Length"
)

# Simplify CDR column names
lengths_long["CDR"] = lengths_long["CDR"].str.replace("_len", "")

# 4) Levene’s test for each organism: do H1, H2, and H3 have equal variances?
print("Levene tests per organism (H1 vs H2 vs H3):\n")
for organism in lengths_long["Source"].unique():
    sub_df = lengths_long[lengths_long["Source"] == organism]
    len_H1 = sub_df[sub_df["CDR"] == "CDR_H1"]["Length"].dropna()
    len_H2 = sub_df[sub_df["CDR"] == "CDR_H2"]["Length"].dropna()
    len_H3 = sub_df[sub_df["CDR"] == "CDR_H3"]["Length"].dropna()
    stat, p = levene(len_H1, len_H2, len_H3)
    print(f"{organism}: W = {stat:.3f}, p = {p:.3g}")
    if p < 0.05:
        print("Significant differences in variance!\n")
    else:
        print("No significant differences.\n")
```

Levene’s test showed highly significant differences in variance across CDR_H1, CDR_H2, and CDR_H3 in all organisms analyzed. This indicates that the variability of CDR lengths is not uniform and that some CDRs fluctuate in length much more than others.



Since Levene’s test checks differences in variances and does not say anything about differences in the median or mean, Kruskal-Wallis and Mann-Whitney test is used to check differences in systematical tendency.
The aim is to answer the question if the medians differ between the three organisms.
In this code, the CDR_H3 sequences from all datasets are loaded and their lengths computed. Then the Kruskal-Wallis test is used to analyse whether at least one group differs significantly in median length. If significant, a pairwise Mann-Whitney U-test is performed to find out which pairs differ.

```python
import pandas as pd
from scipy.stats import kruskal, mannwhitneyu
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Read TSV files
sars_df = pd.read_csv(r'generated\cdrs\seq\corona_cdr_seq.tsv', sep="\t")
human_df = pd.read_csv(r'generated\cdrs\seq\human_cdr_seq.tsv', sep="\t")
influenza_df = pd.read_csv(r'generated\cdrs\seq\influenza_cdr_seq.tsv', sep="\t")

# 2) Extract CDR_H3 lengths into a combined DataFrame
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Prepare lists of lengths per organism
h3_flu = df_all[df_all["Source"] == "Flu"]["CDR_H3_len"].dropna()
h3_human = df_all[df_all["Source"] == "Human"]["CDR_H3_len"].dropna()
h3_sars = df_all[df_all["Source"] == "SARS"]["CDR_H3_len"].dropna()

# 4) Kruskal-Wallis test for differences in medians across all groups
stat, p = kruskal(h3_flu, h3_human, h3_sars)
print(f"Kruskal-Wallis test: H = {stat:.3f}, p = {p:.3g}")

if p < 0.05:
    print("Significant difference in median lengths between groups!")
else:
    print("No significant difference in median lengths.")

# 5) Post-hoc pairwise Mann-Whitney U-tests
print("\nPost-hoc pairwise Mann-Whitney U-tests:")

pairs = [
    ("Flu", h3_flu, "Human", h3_human),
    ("Flu", h3_flu, "SARS", h3_sars),
    ("Human", h3_human, "SARS", h3_sars)
]

for (name1, group1, name2, group2) in pairs:
    u_stat, p_mw = mannwhitneyu(group1, group2, alternative='two-sided')
    print(f"{name1} vs {name2}: U = {u_stat}, p = {p_mw:.3g}")
```

The Kruskal-Wallis test showed a highly significant difference in the median CDR_H3 lengths among the three organisms  p < 0,05.
The post-hoc Mann-Whitney U-tests confirmed that all pairs of groups differ significantly from each other

CDR_H1 and CDR_H2 are also tested in following code

```python
import pandas as pd
from scipy.stats import kruskal, mannwhitneyu
import seaborn as sns
import matplotlib.pyplot as plt

# 1) Load TSV files
sars_df = pd.read_csv(r'generated\cdrs\seq\corona_cdr_seq.tsv', sep="\t")
human_df = pd.read_csv(r'generated\cdrs\seq\human_cdr_seq.tsv', sep="\t")
influenza_df = pd.read_csv(r'generated\cdrs\seq\influenza_cdr_seq.tsv', sep="\t")

# 2) Extract lengths of CDR_H1 and CDR_H2
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H1_len": df["CDR_H1"].str.len(),
        "CDR_H2_len": df["CDR_H2"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Prepare lists of lengths per group
# CDR_H1
h1_flu = df_all[df_all["Source"] == "Flu"]["CDR_H1_len"].dropna()
h1_human = df_all[df_all["Source"] == "Human"]["CDR_H1_len"].dropna()
h1_sars = df_all[df_all["Source"] == "SARS"]["CDR_H1_len"].dropna()

# CDR_H2
h2_flu = df_all[df_all["Source"] == "Flu"]["CDR_H2_len"].dropna()
h2_human = df_all[df_all["Source"] == "Human"]["CDR_H2_len"].dropna()
h2_sars = df_all[df_all["Source"] == "SARS"]["CDR_H2_len"].dropna()

# 4) Kruskal-Wallis test for CDR_H1
print("\nCDR_H1 Kruskal-Wallis test:")
stat_h1, p_h1 = kruskal(h1_flu, h1_human, h1_sars)
print(f"H = {stat_h1:.3f}, p = {p_h1:.3g}")

if p_h1 < 0.05:
    print("Significant difference in median lengths between groups for CDR_H1")
else:
    print("No significant difference in median lengths for CDR_H1.")

# 5) Post-hoc Mann-Whitney U-tests for CDR_H1
print("\nPost-hoc pairwise Mann-Whitney U-tests for CDR_H1:")

pairs_h1 = [
    ("Flu", h1_flu, "Human", h1_human),
    ("Flu", h1_flu, "SARS", h1_sars),
    ("Human", h1_human, "SARS", h1_sars)
]

for (name1, group1, name2, group2) in pairs_h1:
    u_stat, p_mw = mannwhitneyu(group1, group2, alternative='two-sided')
    print(f"{name1} vs {name2}: U = {u_stat}, p = {p_mw:.3g}")

# 6) Kruskal-Wallis test for CDR_H2
print("\nCDR_H2 Kruskal-Wallis test:")
stat_h2, p_h2 = kruskal(h2_flu, h2_human, h2_sars)
print(f"H = {stat_h2:.3f}, p = {p_h2:.3g}")

if p_h2 < 0.05:
    print("Significant difference in median lengths between groups for CDR_H2!")
else:
    print("No significant difference in median lengths for CDR_H2.")

# 7) Post-hoc Mann-Whitney U-tests for CDR_H2
print("\nPost-hoc pairwise Mann-Whitney U-tests for CDR_H2:")

pairs_h2 = [
    ("Flu", h2_flu, "Human", h2_human),
    ("Flu", h2_flu, "SARS", h2_sars),
    ("Human", h2_human, "SARS", h2_sars)
]

for (name1, group1, name2, group2) in pairs_h2:
    u_stat, p_mw = mannwhitneyu(group1, group2, alternative='two-sided')
    print(f"{name1} vs {name2}: U = {u_stat}, p = {p_mw:.3g}")
```

CDR_H1
The Flu sequences have median CDR_H1 lengths distinctly different from both Human and SARS, but Human and SARS do not differ significantly from each other.

CDR_H2
Human sequences differ significantly from both Flu and SARS.
Flu and SARS do not differ significantly from each other.

These findings indicate that CDR length distributions vary systematically between the organisms, but the patterns are not uniform across different CDRs.
