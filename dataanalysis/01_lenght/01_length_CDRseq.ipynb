## CDR amino acid sequence length 

Analysis of CDR amino acid sequence lengths in antibody regions H1, H2, and H3 across coronavirus, human, and influenza.

This Jupyter notebook aims to characterize the lengths of the CDR regions, compare differences between CDR H1, H2, and H3, and identify any variations among antibodies from coronavirus, human, and influenza sources.
Therefore, the lengths of the CDR regions for each organism were first extracted. Additionally, correlations between the lengths of different regions were analyzed to assess whether, for example, a longer CDR H1 sequence is associated with a longer CDR H2 region.

## Coronavirus

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

In this step, a dataset containing SARS-CoV-2 antibody CDR sequences was loaded from a TSV file (corona_cdr_seq.tsv). For each of the three CDR regions (CDR_H1, CDR_H2, and CDR_H3), the code computed the length of the amino acid sequence. Specifically, for every CDR column, a function was applied that determined the number of residues in the sequence if the value was a string; otherwise, a length of 0 was assigned.
The resulting sequence lengths were stored in three new columns named len_H1, len_H2, and len_H3. Finally, these columns were printed to display the lengths of each CDR region for all records in the dataset.


```python
data_corona = pd.read_csv(r"C:\Users\clara\group04-team02\generated\cdrs\seq\corona_cdr_seq.tsv", sep='\t')
for i in range(1,4):
    data_corona[f'len_H{i}'] = data_corona[f'CDR_H{i}'].apply(lambda x: len(x) if isinstance(x, str) else 0)
print(data_corona[['len_H1', 'len_H2', 'len_H3']])
```

This code creates three scatter plots to visually compare the lengths of different CDR regions in the coronavirus antibody dataset:
The first plot shows the relationship between the length of CDR_H1 and CDR_H2.
The second plot compares the lengths of CDR_H2 and CDR_H3.
The third plot compares the lengths of CDR_H1 and CDR_H3.
Each plot uses transparency (alpha=0.5) to help visualize overlapping points better



```python
plt.figure(figsize=(10, 6))
plt.scatter(data_corona['len_H1'], data_corona['len_H2'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H2')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H2')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data_corona['len_H2'], data_corona['len_H3'], alpha=0.5)
plt.title('Length of CDR_H2 vs CDR_H3')
plt.xlabel('Length of CDR_H2')
plt.ylabel('Length of CDR_H3')
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(data_corona['len_H1'], data_corona['len_H3'], alpha=0.5)
plt.title('Length of CDR_H1 vs CDR_H3')
plt.xlabel('Length of CDR_H1')
plt.ylabel('Length of CDR_H3')
plt.show()
```

This code calculates the Pearson correlation coefficients between the lengths of three CDR regions (len_H1, len_H2, len_H3) in the data_corona dataset. Then, it visualizes these correlations as a heatmap with annotations and a color gradient.

```python
corr = data_corona[['len_H1', 'len_H2', 'len_H3']].corr(method='pearson')
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
```

# Comparing SARS-CoV-2, Homo Sapiens and Influenza

The aim in this is to visualize the distribution of lengths of each CDR (CDR_H1, CDR_H2, CDR_H3) per organism in collumns. For each CDR region, it creates a histogram comparing the length distributions of antibodies targeting SARS-CoV-2, humans, and Influenza A. Each group is visualized in a distinct color to easily compare how CDR lengths vary between organisms.

```python
# Make sure all length columns exist:
for df in [data_corona, data_human, data_influenza]:
    for i in range(1, 4):
        col = f'len_H{i}'
        if col not in df.columns:
            df[col] = df[f'CDR_H{i}'].apply(lambda x: len(x) if isinstance(x, str) else 0)

# Define colors
palette = {
    'SARS-CoV-2': '#9509BB',
    'Homo Sapiens': '#FFCF01',
    'Influenza A': '#FF474B'
}

# CDRs
cdrs = ['len_H1', 'len_H2', 'len_H3']

# Plot a histogram for each CDR length
for cdr in cdrs:
    plt.figure(figsize=(10, 6))
    max_len = max(
        data_corona[cdr].max(),
        data_human[cdr].max(),
        data_influenza[cdr].max()
    )
    bins = range(0, max_len + 2)

    plt.hist(
        data_corona[cdr], bins=bins,
        alpha=0.5, label='SARS-CoV-2', color=palette['SARS-CoV-2'], edgecolor='black'
    )
    plt.hist(
        data_human[cdr], bins=bins,
        alpha=0.5, label='Homo Sapiens', color=palette['Homo Sapiens'], edgecolor='black'
    )
    plt.hist(
        data_influenza[cdr], bins=bins,
        alpha=0.5, label='Influenza A', color=palette['Influenza A'], edgecolor='black'
    )

    plt.title(f'Length Distribution of {cdr}')
    plt.xlabel('CDR Length')
    plt.ylabel('Frequency')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

This code creates a boxplot that compares the length distributions of different CDR regions across datasets. Each dataset—SARS-CoV-2, human, and influenza. Compared to the previous step, where the distribution was shown on a histogram, the advantage of a boxplot is that one recognizes more easily differences in distribution and length, with clear and direct comparison to other organsim or CDR region.
The first step to create a grouped boxplot, is to combine all the data in a single table, where each row contains information with CDR region, which dataset it comes from (SARS-CoV-2, Human, or Influenza), and its length.

This reshaping is necessary because Seaborn expects the data in this structure to plot different groups and categories correctly.

One the 'all_data' DataFrame is loaded, Seaborn is ready for plotting.

```python
# 'Dataset' column added to each DataFrame
data_corona['Dataset'] = 'SARS-CoV-2'
data_human['Dataset'] = 'Homo Sapiens'
data_influenza['Dataset'] = 'Influenza A'

# Combine into long format
all_data = pd.concat([
    data_corona.melt(
        id_vars=['Dataset'],
        value_vars=['len_H1', 'len_H2', 'len_H3'],
        var_name='CDR',
        value_name='Length'
    ),
    data_human.melt(
        id_vars=['Dataset'],
        value_vars=['len_H1', 'len_H2', 'len_H3'],
        var_name='CDR',
        value_name='Length'
    ),
    data_influenza.melt(
        id_vars=['Dataset'],
        value_vars=['len_H1', 'len_H2', 'len_H3'],
        var_name='CDR',
        value_name='Length'
    )
], ignore_index=True)
```

```python
sns.boxplot(
    data=all_data,
    x='CDR',
    y='Length',
    hue='Dataset',
    palette={
        'SARS-CoV-2': '#9509BB',
        'Homo Sapiens': '#FFCF01',
        'Influenza A': '#FF474B' 
    }
)
```

Based on this plot and the histograms, it is immediately noticeable that CDR H3 is longer than the other CDRs in the heavy chain, including showing greater variability. Among the CDR H3 regions, Influenza A displays the longest sequences overall. However, when looking at the histogram, it becomes clear that the frequency of these long sequences is very low, suggesting that the significance of this potential outlier still needs to be evaluated. In contrast to H3, CDR H1 and H2 appear to have more similar length distributions, with only a few outliers.

Because CDR lengths are often not normally distributed and can include outliers, as seen in the boxplot, this step calculates the median length for each CDR region and dataset rather than the mean, since the median is less sensitive to extreme values. The code groups all sequences by CDR type and organism, computes the median length in each group, and visualizes these medians in a bar plot. This makes it easier to compare the typical lengths of each CDR region across datasets in a clear and robust way.

```python
# calculating median of each group
median_lengths = all_data.groupby(['CDR', 'Dataset'])['Length'].median().reset_index()

plt.figure(figsize=(8,6))
sns.barplot(
    data=median_lengths,
    x="CDR",
    y="Length",
    hue="Dataset",
    palette = {'SARS-CoV-2': '#9509BB', 'Homo Sapiens': '#FFCF01', 'Influenza A': '#FF474B'}
)
plt.title("Median-Längen der CDRs pro Datensatz")
plt.ylabel("Median CDR-Länge")
plt.xlabel("CDR")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import levene

# Beispiel-Daten einlesen
# (ersetze durch deinen Pfad oder DataFrame)
sars_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\sars_cov2_cdr_seq.tsv", sep="\t")
human_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\human_cdr_seq.tsv", sep="\t")
influenza_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\influenza_cdr_seq.tsv", sep="\t")

# Alle in einen DataFrame packen
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H1_len": df["CDR_H1"].str.len(),
        "CDR_H2_len": df["CDR_H2"].str.len(),
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
])

# Für Levene-Test brauchen wir Listen der Längen
len_H1 = df_all["CDR_H1_len"].dropna()
len_H2 = df_all["CDR_H2_len"].dropna()
len_H3 = df_all["CDR_H3_len"].dropna()

# Levene-Test
stat, p_value = levene(len_H1, len_H2, len_H3)
print(f"Levene-Test: W = {stat:.3f}, p = {p_value:.3g}")

# Interpretation
if p_value < 0.05:
    print("➡️ Signifikant unterschiedliche Varianzen!")
else:
    print("➡️ Keine signifikanten Unterschiede in der Varianz.")

# Boxplot zur Visualisierung
df_melt = df_all.melt(
    value_vars=["CDR_H1_len", "CDR_H2_len", "CDR_H3_len"],
    var_name="CDR",
    value_name="Length"
)

plt.figure(figsize=(8,6))
sns.boxplot(x="CDR", y="Length", data=df_melt)
plt.title("Längenverteilung der CDRs (H1, H2, H3)")
plt.ylabel("CDR Länge")
plt.xlabel("CDR Typ")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import levene

# 1) Daten einlesen
sars_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\sars_cov2_cdr_seq.tsv", sep="\t")
human_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\human_cdr_seq.tsv", sep="\t")
influenza_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\influenza_cdr_seq.tsv", sep="\t")

# 2) Längen extrahieren
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H1_len": df["CDR_H1"].str.len(),
        "CDR_H2_len": df["CDR_H2"].str.len(),
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Long Format
lengths_long = df_all.melt(
    id_vars="Source",
    value_vars=["CDR_H1_len", "CDR_H2_len", "CDR_H3_len"],
    var_name="CDR",
    value_name="Length"
)

# CDR Namen vereinfachen
lengths_long["CDR"] = lengths_long["CDR"].str.replace("_len", "")

# 4) Levene-Test pro Organismus
print("Levene-Tests pro Organismus (H1 vs H2 vs H3):\n")
for organism in lengths_long["Source"].unique():
    sub_df = lengths_long[lengths_long["Source"] == organism]
    len_H1 = sub_df[sub_df["CDR"] == "CDR_H1"]["Length"].dropna()
    len_H2 = sub_df[sub_df["CDR"] == "CDR_H2"]["Length"].dropna()
    len_H3 = sub_df[sub_df["CDR"] == "CDR_H3"]["Length"].dropna()
    stat, p = levene(len_H1, len_H2, len_H3)
    print(f"{organism}: W = {stat:.3f}, p = {p:.3g}")
    if p < 0.05:
        print("➡️ Signifikant unterschiedliche Varianzen!\n")
    else:
        print("➡️ Keine signifikanten Unterschiede.\n")

# 5) Boxplot nach Organismus
plt.figure(figsize=(10,6))
sns.boxplot(x="CDR", y="Length", hue="Source", data=lengths_long)
plt.title("Längenverteilung CDR_H1, CDR_H2, CDR_H3 pro Organismus")
plt.ylabel("CDR Länge")
plt.xlabel("CDR Typ")
plt.legend(title="Organismus")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from scipy.stats import kruskal, mannwhitneyu
import seaborn as sns
import matplotlib.pyplot as plt

# 1) TSV-Dateien einlesen
sars_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\sars_cov2_cdr_seq.tsv", sep="\t")
human_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\human_cdr_seq.tsv", sep="\t")
influenza_df = pd.read_csv(r"C:\Users\clara\group04-team02\data cleanup\influenza_cdr_seq.tsv", sep="\t")

# 2) Alle in einen DataFrame packen
def extract_lengths(df, label):
    return pd.DataFrame({
        "CDR_H3_len": df["CDR_H3"].str.len(),
        "Source": label
    })

df_all = pd.concat([
    extract_lengths(sars_df, "SARS"),
    extract_lengths(human_df, "Human"),
    extract_lengths(influenza_df, "Flu")
], ignore_index=True)

# 3) Listen für den Test
h3_flu = df_all[df_all["Source"]=="Flu"]["CDR_H3_len"].dropna()
h3_human = df_all[df_all["Source"]=="Human"]["CDR_H3_len"].dropna()
h3_sars = df_all[df_all["Source"]=="SARS"]["CDR_H3_len"].dropna()

# 4) Kruskal-Wallis-Test
stat, p = kruskal(h3_flu, h3_human, h3_sars)
print(f"Kruskal-Wallis-Test: H = {stat:.3f}, p = {p:.3g}")

if p < 0.05:
    print("➡️ Signifikanter Unterschied in den Median-Längen zwischen den Gruppen!")
else:
    print("➡️ Kein signifikanter Unterschied in den Median-Längen.")

# 5) Post-hoc Mann-Whitney-Tests
print("\nPost-hoc Paarweise Mann-Whitney-Tests:")

pairs = [
    ("Flu", h3_flu, "Human", h3_human),
    ("Flu", h3_flu, "SARS", h3_sars),
    ("Human", h3_human, "SARS", h3_sars)
]

for (name1, group1, name2, group2) in pairs:
    u_stat, p_mw = mannwhitneyu(group1, group2, alternative='two-sided')
    print(f"{name1} vs {name2}: U = {u_stat}, p = {p_mw:.3g}")

# 6) Boxplot zur Visualisierung
plt.figure(figsize=(8,6))
sns.boxplot(x="Source", y="CDR_H3_len", data=df_all, palette="Set2")
plt.title("CDR-H3 Längen nach Organismus")
plt.xlabel("Organismus")
plt.ylabel("CDR-H3 Länge")
plt.tight_layout()
plt.show()
```
